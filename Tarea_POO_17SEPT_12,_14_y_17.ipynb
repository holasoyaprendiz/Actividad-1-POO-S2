```python
#ACTIVIDAD #12
class actividad12:
  @staticmethod
  def calcular_salarioB (horas_trabajadas,valor_hora):
    return horas_trabajadas * valor_hora
  @staticmethod
  def calcular_retencionF (retencion,salarioB):
    return retencion * salarioB
  @staticmethod
  def calcular_salarioN (salarioB,retencionF):
    return salarioB - retencionF




#Variables
horas_trabajadas = 48
valor_hora = 5000
retencion = 0.125
#calculos usando la clase
salarioB = actividad12.calcular_salarioB(horas_trabajadas,valor_hora)
retencionF = actividad12.calcular_retencionF(retencion,salarioB)
salarioN = actividad12.calcular_salarioN (salarioB,retencionF)

#imprimir
print(f"El salario bruto es {salarioB} ")
print(f"El salario neto es {salarioN} ")
print(f"La retención en la fuente es {retencionF} ")
```

```python
#ACTIVIDAD #14
import math
class potencia:
  @staticmethod
  def calcular_cuadrado (num):
    return math.pow(num,2)
  @staticmethod
  def calcular_cubo (num):
    return math.pow(num,3)

 #definir variables
num = int(input())
cuadrado = potencia.calcular_cuadrado(num)
cubo = potencia.calcular_cubo(num)

#imprimir
print(f"el cuadrado del número es: {cuadrado}")
print(f"el cubo del número es: {cubo}")
```

```python
#ACTIVIDAD #17
import math
class circle:
  @staticmethod
  def calcular_area(radio):
    return math.pi*math.pow(radio,2)
  @staticmethod
  def calcular_longitud (radio):
    return radio * 2 * math.pi

#definir variables
radio = int(input())
longitud = circle.calcular_longitud(radio)
area = circle.calcular_area(radio)
#imprimir
print(f"El área del circulo es {area}")
print(f"La longitud del circulo es {longitud}")
```
